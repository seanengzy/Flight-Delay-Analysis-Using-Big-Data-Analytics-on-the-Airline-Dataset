```python
import matplotlib.pyplot as plt
```

Section 1 - Data Loading

```python
methods = ["PySpark", "Pandas", "SparkSQL", "Local SQL"]

times = [
    129.117,   # PySpark
    229.198,   # Pandas
    325.259,   # SparkSQL
    116.967    # Local SQL
]

plt.figure(figsize=(8, 5))

bars = plt.bar(methods, times)

plt.title("Data Loading Execution Time Comparison")
plt.xlabel("Method")
plt.ylabel("Execution Time (Seconds)")
plt.grid(axis="y", alpha=0.3)

for bar, time in zip(bars, times):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{time:.3f} s",
        ha="center",
        va="bottom"
    )

plt.ylim(0, max(times) * 1.15)
plt.tight_layout()
plt.show()
```

Total Execution Time Comparison

```python
methods = ["PySpark", "Pandas", "SparkSQL", "Local SQL"]

times = [
    353.312,   # PySpark
    612.026,   # Pandas
    396.925,   # SparkSQL
    613.729    # Local SQL
]

plt.figure(figsize=(8, 5))

bars = plt.bar(methods, times)

plt.title("Total Execution Time Comparison")
plt.xlabel("Method")
plt.ylabel("Total Execution Time (Seconds)")
plt.grid(axis="y", alpha=0.3)

for bar, time in zip(bars, times):

    minutes = time / 60

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{time:.3f} s\n({minutes:.2f} min)",
        ha="center",
        va="bottom"
    )

plt.ylim(0, max(times) * 1.15)
plt.tight_layout()
plt.show()
```
